### Key Equations from IRHv57.md:

**7.1 Self-Dual Cancellation:**
$$E_{vac} = \sum (\text{Boson Modes}) - \sum (\text{Fermion Modes})$$

**7.2 Geometric Deficit:**
- Bosonic sites vs Fermionic holes
- Self-duality: E_site + E_hole ≈ 0
- Horizon limit creates residual

**7.3 Cosmological Constant:**
$$\Lambda \approx \frac{\delta}{R_H^2}$$

where:
- δ = geometric deficit at horizon
- R_H = Hubble radius

**Derivation Strategy:**
1. Calculate naive QFT vacuum energy
2. Apply self-dual cancellation mechanism
3. Derive diffraction limit at horizon
4. Calculate residual Λ
5. Validate against Planck 2018 data

```python
# Cell 3: Symbolic Derivation - Self-Dual Cancellation

print("\n" + "="*60)
print("STEP 1: Naive QFT Vacuum Energy")
print("="*60)

print("\nStandard QFT prediction:")
print("  ρ_vac ~ M_P⁴ ~ 10¹²³ × ρ_observed")
print("  The worst prediction in physics!")

print("\n" + "="*60)
print("STEP 2: D₄ Self-Duality Mechanism")
print("="*60)

print("\nD₄ lattice has dual structure:")
print("  • Bosonic sites: E_boson > 0")
print("  • Fermionic holes (Voronoi): E_fermion < 0")
print("\nSelf-duality (Λ ≅ Λ*):")
print("  Number of sites = Number of holes")
print("  E_site + E_hole ≈ 0 (cancellation!)")

print("\n" + "="*60)
print("STEP 3: Horizon Diffraction Limit")
print("="*60)

# Define symbols
R_H, a0, delta = symbols('R_H a_0 delta', positive=True, real=True)

print("\nPerfect cancellation only in infinite lattice.")
print("\nUniverse is finite (Hubble radius R_H):")
print("  Boundary effects create geometric deficit δ")
print("\nDiffraction limit:")
print("  Δk ~ 1/R_H")
print("  Residual energy density: ρ_Λ ~ 1/R_H⁴")

# Cosmological constant
Lambda_symbolic = delta / R_H**2

print("\nCosmological constant:")
print(f"  Λ = δ/R_H²")
print(f"  Geometric deficit over horizon area")

print("\n✓ Λ is small NOW because R_H is large!")
print("  Λ scales with universe size (not constant)")
```

```python
# Cell 4: Numerical Computation

print("\n" + "="*60)
print("STEP 4: Numerical Evaluation")
print("="*60)

# Physical constants
c_val = constants.c
hbar_val = constants.hbar
G_val = constants.G

# Planck scales
M_P = np.sqrt(hbar_val * c_val / G_val)
L_P = np.sqrt(hbar_val * G_val / c_val**3)
rho_P = M_P / L_P**3

print(f"\nPlanck scales:")
print(f"  M_P = {M_P:.5e} kg")
print(f"  L_P = {L_P:.5e} m")
print(f"  ρ_P = {rho_P:.5e} kg/m³")

# Hubble radius (current epoch)
H_0 = 67.4 * 1e3 / (3.086e22)  # Planck 2018 (s⁻¹)
R_H_val = c_val / H_0

print(f"\nHubble parameters:")
print(f"  H₀ = {H_0:.5e} s⁻¹")
print(f"  R_H = {R_H_val:.5e} m")
print(f"  R_H = {R_H_val/L_P:.5e} L_P")

# Geometric deficit estimate
# δ ~ (lattice spacing)² / (horizon area)
delta_val = L_P**2 / R_H_val**2

print(f"\nGeometric deficit:")
print(f"  δ ~ L_P²/R_H² = {delta_val:.5e}")

# Cosmological constant
Lambda_val = delta_val / R_H_val**2

print(f"\nCosmological constant:")
print(f"  Λ_theory = δ/R_H² = {Lambda_val:.5e} m⁻²")

# Energy density
rho_Lambda_val = Lambda_val * c_val**2 / (8 * np.pi * G_val)

print(f"  ρ_Λ = {rho_Lambda_val:.5e} kg/m³")

# Store results
results = {
    'M_P': float(M_P),
    'L_P': float(L_P),
    'R_H': float(R_H_val),
    'delta': float(delta_val),
    'Lambda': float(Lambda_val),
    'rho_Lambda': float(rho_Lambda_val)
}
```
